```python
import requests

url = "http://localhost:8000/api/documents/upload"

files = {
    "file": ("EJEMPLO DE CONTRATO - RETO 1.pdf", open("c:\\EJEMPLO DE CONTRATO - RETO 1.pdf", "rb"), "application/pdf")
}

response = requests.post(url, files=files)

print("Status:", response.status_code)
print("Response:", response.json())
```

```python
session_id = response.json().get("data", {}).get("session_id")
print("Session ID:", session_id)
```

```python
url = "http://127.0.0.1:8000/api/analysis/tender"

payload = {
    "analysis_type": "comprehensive",   # o "compliance_only", etc.
    "session_id": session_id,  # usa el session_id que obtuviste al subir el archivo
    # "conversation_id": str(uuid.uuid4())  # opcional, si quieres mandar uno
}

headers = {
    "Content-Type": "application/json",
    "Accept": "application/json"
}

response = requests.post(url, json=payload, headers=headers)

print("Status:", response.status_code)
print("Response:", response.json())
```
